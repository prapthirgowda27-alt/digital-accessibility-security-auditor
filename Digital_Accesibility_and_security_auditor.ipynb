```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse

url = input("Enter website URL: ").strip()

parsed = urlparse(url)

if parsed.scheme not in ["http", "https"] or not parsed.hostname:
    print("Invalid URL")
else:
    try:
        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "AccessibilityStudentAuditor/1.0"}
        )

        response.raise_for_status()
        soup = BeautifulSoup(response.text, "html.parser")

        images = soup.find_all("img")
        missing_alt = []

        for image in images:
            if image.get("alt") is None:
                missing_alt.append(image)

        print("\n===== ALT TEXT ACCESSIBILITY REPORT =====")
        print("Total images:", len(images))
        print("Images without ALT:", len(missing_alt))
        print("Images with ALT:", len(images) - len(missing_alt))

        if missing_alt:
            print("\nImages requiring review:")
            for i, image in enumerate(missing_alt, 1):
                print(i, image.get("src", "Source unavailable"))
        else:
            print("No images missing ALT attributes.")

    except requests.RequestException as error:
        print("Error:", error)
```

```python

import requests
from bs4 import BeautifulSoup

url = "https://www.india.gov.in/"

response = requests.get(url, timeout=15)
soup = BeautifulSoup(response.text, "html.parser")

links = soup.find_all("a")

missing_names = []

for link in links:

    text = link.get_text(strip=True)
    aria_label = link.get("aria-label")
    title = link.get("title")

    images = link.find_all("img")
    image_alt = any(
        img.get("alt", "").strip()
        for img in images
    )

    if not text and not aria_label and not title and not image_alt:
        missing_names.append(link)

print("===== LINK ACCESSIBILITY REPORT =====")

print("Total links:", len(links))
print("Links without accessible names:", len(missing_names))

for i, link in enumerate(missing_names, 1):
    print(i, link.get("href", "No URL"))

print("\nLink analysis completed!")
```

```python

import requests
from bs4 import BeautifulSoup

url = "https://www.india.gov.in/"

response = requests.get(url, timeout=15)
soup = BeautifulSoup(response.text, "html.parser")

controls = soup.find_all(["input", "textarea", "select"])

missing_labels = []

for control in controls:

    control_type = control.get("type", "").lower()

    if control_type in ["hidden", "submit", "button", "reset", "image"]:
        continue

    control_id = control.get("id")

    has_label = False

    if control_id:
        label = soup.find("label", attrs={"for": control_id})
        if label:
            has_label = True

    if control.find_parent("label"):
        has_label = True

    if control.get("aria-label"):
        has_label = True

    if control.get("aria-labelledby"):
        has_label = True

    if not has_label:
        missing_labels.append(control)

print("===== FORM ACCESSIBILITY REPORT =====")

print("Total form controls:", len(controls))
print("Controls without labels:", len(missing_labels))

for i, control in enumerate(missing_labels, 1):
    print(
        i,
        control.name,
        control.get("type", "N/A"),
        control.get("name", "Unnamed")
    )

print("\nForm accessibility analysis completed!")
```

```python


import gradio as gr
import requests
import socket
import ipaddress
from bs4 import BeautifulSoup
from urllib.parse import urlparse

import os
import html
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.platypus import Table, TableStyle
from reportlab.platypus import KeepTogether
from reportlab.lib import colors

def scan_website(url):

    url = url.strip()
    parsed = urlparse(url)

    if parsed.scheme not in ["http", "https"] or not parsed.hostname:
        return "Please enter a valid URL starting with https://", None
    try:
        # Basic protection: only scan public website addresses
        hostname = parsed.hostname
        addresses = socket.getaddrinfo(hostname, None)

        for address in addresses:
            ip = ipaddress.ip_address(address[4][0])
            if not ip.is_global:
                return "Private or local network addresses are not allowed.", None

        response = requests.get(
            url,
            timeout=15,
            allow_redirects=False,
            headers={"User-Agent": "StudentAccessibilityAuditor/1.0"}
        )

        if 300 <= response.status_code < 400:
            return "Website redirected. Please enter its final HTTPS URL."

        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")
        headers = response.headers

        # IMAGE ANALYSIS
        images = soup.find_all("img")
        missing_alt = [
            img for img in images if not img.has_attr("alt")
        ]
        empty_alt = [
            img for img in images
            if img.has_attr("alt") and not img["alt"].strip()
        ]


        # LINK ANALYSIS
        links = soup.find_all("a")
        missing_links = []

        for link in links:
            text = link.get_text(strip=True)
            aria = link.get("aria-label", "").strip()
            title = link.get("title", "").strip()

            img_alt = any(
                img.get("alt", "").strip()
                for img in link.find_all("img")
            )

            if not (text or aria or title or img_alt):
                missing_links.append(link)

        # FORM ANALYSIS
        controls = soup.find_all(["input", "textarea", "select"])
        missing_labels = []

        for control in controls:
            control_type = control.get("type", "").lower()

            if control_type in ["hidden", "submit", "button", "reset", "image"]:
                continue

            control_id = control.get("id")
            labelled = False

            if control_id and soup.find(
                "label", attrs={"for": control_id}
            ):
                labelled = True

            if control.find_parent("label"):
                labelled = True

            if control.get("aria-label") or control.get("aria-labelledby"):
                labelled = True

            if control_id and soup.find(
                attrs={"aria-labelledby": control_id}
            ):
                labelled = True

            if not labelled:
                missing_labels.append(control)
                print("Unlabelled control:", control)



        # HEADING ANALYSIS
        heading_counts = {
            f"H{i}": len(soup.find_all(f"h{i}"))
            for i in range(1, 7)
        }

        # SECURITY ANALYSIS

        security_headers = {
            "Content-Security-Policy": "CSP",
            "X-Content-Type-Options": "MIME Sniffing Protection",
            "Strict-Transport-Security": "HSTS"
        }

        present = [
            name for name in security_headers
            if name in headers
        ]

        csp_value = headers.get("Content-Security-Policy", "")
        x_frame_value = headers.get("X-Frame-Options", "")

        clickjacking_protection = (
            "frame-ancestors" in csp_value.lower()
            or bool(x_frame_value)
        )

        if clickjacking_protection:
            present.append("Clickjacking Protection")

        missing_security = [
            label for name, label in security_headers.items()
            if name not in headers
        ]

        if not clickjacking_protection:
            missing_security.append("Clickjacking Protection")


        # BASIC PROTOTYPE SCORES
        total_checks = (
            len(images) + len(links) + len(controls)
        )

        issues = (
            len(missing_alt) +
            len(missing_links) +
            len(missing_labels)
        )

        accessibility_score = (
            max(0, (total_checks - issues) / total_checks * 100)
            if total_checks > 0 else None
        )

        security_score = (
            (len(present) + int(url.startswith("https://"))) / 5 * 100
        )

        # FINAL REPORT
        report = f"""
# DIGITAL ACCESSIBILITY AUDIT REPORT

**Website:** {response.url}

**HTTP Status:** {response.status_code}

## 1. Accessibility Results

| Test | Result |
|---|---:|
| Total images | {len(images)} |
| Missing ALT text | {len(missing_alt)} |
| Empty ALT text | {len(empty_alt)} |
| Total links | {len(links)} |
| Links without detected names | {len(missing_links)} |
| Form controls | {len(controls)} |
| Controls without detected labels | {len(missing_labels)} |

### Heading Structure

| Heading | Count |
|---|---:|
| H1 | {heading_counts['H1']} |
| H2 | {heading_counts['H2']} |
| H3 | {heading_counts['H3']} |
| H4 | {heading_counts['H4']} |
| H5 | {heading_counts['H5']} |
| H6 | {heading_counts['H6']} |

**Prototype Accessibility Score: {f'{accessibility_score:.1f}/100' if accessibility_score is not None else 'Not enough data to calculate'}**

## 2. Security Results

**HTTPS:** {"Present" if url.startswith("https://") else "Not used"}

| Security check | Result |
|---|---|
"""
                        for name, label in security_headers.items():
            status = "Present" if name in headers else "Not detected"
            report += f"| {label} | {status} |\n"

        clickjacking_status = (
            "Present" if clickjacking_protection else "Not detected"
        )

        report += f"| Clickjacking Protection | {clickjacking_status} |\n"
                report += f"""

## 3. Recommendations

"""
        recommendations = []

        if missing_alt:
            recommendations.append("Review images missing ALT attributes.")

        if empty_alt:
            recommendations.append(
                "Review empty ALT attributes; decorative images may correctly use empty ALT."
            )

        if missing_links:
            recommendations.append("Review links without detected accessible names.")

        if missing_labels:
            recommendations.append("Review form controls without detected labels.")

        if missing_security:
            recommendations.append(
                "Review missing security headers: " +
                ", ".join(missing_security)
            )

        if not url.startswith("https://"):
            recommendations.append("Use HTTPS.")

        if not recommendations:
            recommendations.append(
                "No issues were detected by these basic checks. "
                "Perform a complete accessibility and security review."
            )

        for item in recommendations:
            report += f"- {item}\n"

        report += """
--
*This is a preliminary automated audit, not official WCAG certification or a complete security assessment. Manual testing is also necessary.*
"""



        pdf_path = "/content/website_audit_report.pdf"

        doc = SimpleDocTemplate(
            pdf_path,
            pagesize=A4,
            rightMargin=40,
            leftMargin=40,
            topMargin=40,
            bottomMargin=40
        )

        styles = getSampleStyleSheet()
        story = []

        story.append(Paragraph(
            "Digital Accessibility & Security Auditor",
            styles["Title"]
        ))
        story.append(Spacer(1, 8))


        table_data = []

        def add_table():
            if table_data:
                formatted_data = [
                    [Paragraph(html.escape(cell), styles["Normal"]) for cell in row]
                    for row in table_data
                ]

                table = Table(formatted_data, repeatRows=1)
                table.setStyle(TableStyle([
                    ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#243B55")),
                    ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
                    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
                    ("PADDING", (0, 0), (-1, -1), 4),
                    ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
                ]))

                story.append(table)
                story.append(Spacer(1, 4))
                table_data.clear()

        for line in report.splitlines():
            stripped = line.strip()

            if stripped.startswith("|") and stripped.endswith("|"):
                cells = [cell.strip() for cell in stripped.strip("|").split("|")]

                if not all(set(cell) <= set("-: ") for cell in cells):
                    table_data.append(cells)
                continue

            add_table()

            if stripped.startswith("### "):
                story.append(Paragraph(html.escape(stripped[4:]), styles["Heading2"]))
            elif stripped.startswith("## "):
                story.append(Paragraph(html.escape(stripped[3:]), styles["Heading1"]))
            elif stripped.startswith("# "):
                story.append(Paragraph(html.escape(stripped[2:]), styles["Heading1"]))
            elif stripped:
                clean_line = stripped.replace("**", "")

                if clean_line.startswith("*This is a preliminary"):
                    clean_line = clean_line.strip("*")
                    story.append(KeepTogether([
                        Paragraph(html.escape(clean_line), styles["Normal"])
                    ]))
                else:
                    story.append(Paragraph(html.escape(clean_line), styles["Normal"]))


            story.append(Spacer(1, 0))

        add_table()


        doc.build(story)

        return report, pdf_path

    except Exception as error:
        return f"Scan failed: {str(error)}", None




# DASHBOARD DESIGN


with gr.Blocks(
    title="Digital Accessibility & Security Auditor",
    theme=gr.themes.Soft()
) as app:

    gr.Markdown("""
    # 🛡️ Digital Accessibility & Security Auditor

    ### Evaluate public websites for accessibility and basic security settings.

    Enter a public website URL below to begin.
    """)

    website_url = gr.Textbox(
        label="Enter Website URL",
        placeholder="https://www.india.gov.in/"
    )

    scan_button = gr.Button(
        "🔍 Scan Website",
        variant="primary"
    )

    results = gr.Markdown(label="Audit Report")

    pdf_download = gr.File(label="Download PDF Audit Report")

    scan_button.click(
        fn=scan_website,
        inputs=website_url,
        outputs=[results, pdf_download]
    )

app.launch(share=True)
```
